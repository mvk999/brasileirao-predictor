```python
df_2024 = df_clean[
    df_clean["temporada"] == 2024
].copy()
```

```python
df_2024["pontos_mandante"] = 0
df_2024["pontos_visitante"] = 0
```

```python
df_2024.loc[
    df_2024["resultado"] == "H",
    "pontos_mandante"
] = 3
```

```python
df_2024.loc[
    df_2024["resultado"] == "A",
    "pontos_visitante"
] = 3
```

```python
df_2024.loc[
    df_2024["resultado"] == "D",
    ["pontos_mandante", "pontos_visitante"]
] = 1
```

```python
df_2024[
    [
        "mandante",
        "visitante",
        "resultado",
        "pontos_mandante",
        "pontos_visitante"
    ]
].head(10)
```

```python
pontos_casa = (
    df_2024
    .groupby("mandante")["pontos_mandante"]
    .sum()
)
```

```python
pontos_fora = (
    df_2024
    .groupby("visitante")["pontos_visitante"]
    .sum()
)
```

```python
tabela_2024 = pd.DataFrame({
    "pontos_casa": pontos_casa,
    "pontos_fora": pontos_fora
})
```

```python
tabela_2024["pontos"] = (
    tabela_2024["pontos_casa"]
    + tabela_2024["pontos_fora"]
)
```

```python
tabela_2024.sort_values(
    "pontos",
    ascending=False
)
```

```python
vitorias_casa = (
    df_2024[df_2024["resultado"] == "H"]
    ["mandante"]
    .value_counts()
)
```

```python
vitorias_fora = (
    df_2024[df_2024["resultado"] == "A"]
    ["visitante"]
    .value_counts()
)
```

```python
tabela_2024["vitorias"] = (
    vitorias_casa.add(vitorias_fora, fill_value=0)
)
```

```python
empates_casa = (
    df_2024[df_2024["resultado"] == "D"]
    ["mandante"]
    .value_counts()
)

empates_fora = (
    df_2024[df_2024["resultado"] == "D"]
    ["visitante"]
    .value_counts()
)
```

```python
tabela_2024["empates"] = (
    empates_casa.add(empates_fora, fill_value=0)
)
```

```python
derrotas_casa = (
    df_2024[df_2024["resultado"] == "A"]
    ["mandante"]
    .value_counts()
)
```

```python
derrotas_fora = (
    df_2024[df_2024["resultado"] == "H"]
    ["visitante"]
    .value_counts()
)
```

```python
tabela_2024["derrotas"] = (
    derrotas_casa.add(derrotas_fora, fill_value=0)
)
```

```python
tabela_2024["jogos"] = (
    tabela_2024["vitorias"]
    + tabela_2024["empates"]
    + tabela_2024["derrotas"]
)
```

```python
tabela_2024[
    ["pontos", "jogos", "vitorias", "empates", "derrotas"]
].sort_values("pontos", ascending=False)
```

```python
gols_pro_casa = (
    df_2024
    .groupby("mandante")["gols_mandante"]
    .sum()
)
```

```python
gols_pro_fora = (
    df_2024
    .groupby("visitante")["gols_visitante"]
    .sum()
)
```

```python
tabela_2024["gols_pro"] = (
    gols_pro_casa.add(gols_pro_fora, fill_value=0)
)
```

```python
df_2024["gols_visitante"]
```

```python
gols_contra_casa = (
    df_2024
    .groupby("mandante")["gols_visitante"]
    .sum()
)
```

```python
gols_contra_fora = (
    df_2024
    .groupby("visitante")["gols_mandante"]
    .sum()
)
```

```python
tabela_2024["gols_contra"] = (
    gols_contra_casa.add(gols_contra_fora, fill_value=0)
)
```

```python
tabela_2024["saldo_gols"] = (
    tabela_2024["gols_pro"]
    - tabela_2024["gols_contra"]
)
```

```python
tabela_2024[
    [
        "pontos",
        "jogos",
        "vitorias",
        "empates",
        "derrotas",
        "gols_pro",
        "gols_contra",
        "saldo_gols"
    ]
].sort_values(
    ["pontos", "vitorias", "saldo_gols", "gols_pro"],
    ascending=False
)
```

```python
tabela_2024["aproveitamento_fora"] = (
    tabela_2024["pontos_fora"] / 57 * 100
).round(2)
```
